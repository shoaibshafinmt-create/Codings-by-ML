**Create the simplest Not function code**

```python
print(not True)
print(not False)
```

**Create a basic type of Not function code**

```python
print("A   |   not A")
print("--------------")
for a in[True, False]:
  print(a, " | ", not a)
print("\nWith 1 and 0:")
print(int(not 1), int(not 0))
```

**Create a very basic Truth table**

```python
print("A     | not A")
print("-------------")
for a in [True, False]:
    print(a, " | ", not a)
```
