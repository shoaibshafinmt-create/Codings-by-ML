**Create the simple Inverter code**

```python
def inverter(input_bit):
    if input_bit == 0:
        return 1
    else:
        return 0
print(inverter(1))
print(inverter(0))
```

**Create the Truth Table**

```python
print("A | B")
print("------")
for A in [0,1]:
  B = inverter(A)
  print(A, " | ", B)
```

**How to make the Inverter with Boolean Values**

```python
print(not A)
print(not B)
```

**How to use a single neuron while using Inverter**

```python
def step(x):
  if x>0:
    return 1
  else:
    return 0
def not_neuron(input_bit):
    Weight = -1
    Bias = 0.5
    total = input_bit* Weight + Bias
    return step(total)
print(not_neuron(0))
print(not_neuron(1))
```
